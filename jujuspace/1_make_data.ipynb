```python
import cv2
import os
import numpy as np
from tqdm import tqdm

# 비디오 경로 설정
video_path = "../data/origin_video.mp4"
output_folder = "../data/origin_images"

# 추출할 구간들 (초 단위) - 여기를 수정하세요
time_ranges = [
    (70.0, 80.0),    # 첫 번째 구간: 70초 ~ 80초
    (90.0, 93.0),   # 두 번째 구간: 90초 ~ 93초
    (268.0, 272.0)    # 세 번째 구간: 268초 ~ 272초
]

# origin_images 폴더 생성
os.makedirs(output_folder, exist_ok=True)

print("비디오에서 여러 구간의 프레임을 추출합니다...")
print(f"비디오 경로: {video_path}")
print(f"저장 폴더: {output_folder}")
print(f"추출 구간들:")
for i, (start, end) in enumerate(time_ranges, 1):
    print(f"  {i}. {start}초 ~ {end}초")

# 비디오 캡처 객체 생성
cap = cv2.VideoCapture(video_path)

if not cap.isOpened():
    print("비디오 파일을 열 수 없습니다!")
else:
    # 비디오 정보 출력
    total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
    fps = cap.get(cv2.CAP_PROP_FPS)
    duration = total_frames / fps
    
    print(f"\n비디오 정보:")
    print(f"총 프레임 수: {total_frames}")
    print(f"FPS: {fps}")
    print(f"총 시간: {duration:.2f}초")
    
    total_saved = 0
    
    # 각 구간별로 프레임 추출
    for segment_idx, (start_time, end_time) in enumerate(time_ranges, 1):
        print(f"\n=== 구간 {segment_idx}: {start_time}초 ~ {end_time}초 ===")
        
        # 시작/끝 프레임 계산
        start_frame = int(start_time * fps)
        end_frame = int(end_time * fps)
        end_frame = min(end_frame, total_frames - 1)  # 범위 제한
        
        if start_frame >= total_frames:
            print(f"구간 {segment_idx}: 시작 시간이 비디오 길이를 초과합니다.")
            continue
            
        print(f"프레임 범위: {start_frame} ~ {end_frame}")
        print(f"추출할 프레임 수: {end_frame - start_frame + 1}")
        
        # 시작 프레임으로 이동
        cap.set(cv2.CAP_PROP_POS_FRAMES, start_frame)
        
        # 지정된 구간의 프레임 추출
        frame_count = start_frame
        saved_count = 0
        
        with tqdm(total=end_frame - start_frame + 1, desc=f"구간 {segment_idx} 추출") as pbar:
            while frame_count <= end_frame:
                ret, frame = cap.read()
                if not ret:
                    break
                
                # 프레임을 이미지 파일로 저장 (구간별로 폴더 분리)
                segment_folder = os.path.join(output_folder, f"segment_{segment_idx}")
                os.makedirs(segment_folder, exist_ok=True)
                
                filename = f"frame_{frame_count:06d}.jpg"
                filepath = os.path.join(segment_folder, filename)
                
                # 이미지 저장
                cv2.imwrite(filepath, frame)
                saved_count += 1
                
                frame_count += 1
                pbar.update(1)
        
        print(f"구간 {segment_idx} 완료: {saved_count}개 프레임 저장")
        total_saved += saved_count
    
    cap.release()
    
    print(f"\n전체 추출 완료!")
    print(f"총 {total_saved}개의 프레임이 저장되었습니다.")
    print(f"저장 위치: {output_folder}/segment_1, segment_2, segment_3")
```
